# Introduccion

Hacer un codigo de python que le dices el día, mes y año y te saca que dia de la semana es. Tabien conocida como congruencia de zeller.

### Actividad: Congruencia de Zeller

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
from datetime import date, timedelta
```

```python
def diaSemana(fecha):
    dia = fecha.day
    mes = fecha.month
    añoTemp = fecha.year

    if mes < 3:
        mes += 12
        añoTemp -= 1

    año = añoTemp %100
    siglo = añoTemp // 100

    h = (dia + ((13 * (mes + 1)) // 5) + año + (año // 4) + (siglo // 4) - (2 * siglo)) % 7

    return h

def diaParser(dia):
    dias = ["sabado","domingo","lunes","martes","miercoles","jueves","viernes"]

    return dias[dia]
```

### Reto: Primera IA

```python
data = []

for i in range(1,4000):
    fecha = date(2026,9,29) + timedelta(days=i)
    diasem = diaSemana(fecha)
    data.append([diaParser(diasem),diasem,fecha.day,fecha.month,fecha.year])

df = pd.DataFrame(data, columns=["Dia semana", "numDiaSemana", "Dia", "Mes", "Año"])
df
```

```python
X = df[["Dia","Mes","Año"]]
Y = df[["numDiaSemana"]]

print(X)
print(Y)

x_train, x_test, y_train, y_test = train_test_split(X,Y,test_size=0.3,random_state=0)

logisticregrression = LogisticRegression()

logisticregrression.fit(x_train,y_train.values.ravel())

predictions = logisticregrression.predict(x_test)
score = logisticregrression.score(x_test,y_test)
params = logisticregrression.get_params(deep=True)

print(score)
```
